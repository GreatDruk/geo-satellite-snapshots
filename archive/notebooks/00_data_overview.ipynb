```python
from pathlib import Path
import sys

REPOSITORY_ROOT = next(
    p for p in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (p / "archive/src/experiment.py").is_file()
)
PROJECT_ROOT = ROOT = REPOSITORY_ROOT / "archive"
if str(REPOSITORY_ROOT) not in sys.path:
    sys.path.insert(0, str(REPOSITORY_ROOT))
```

```python
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
DATA_DIR: Path = REPOSITORY_ROOT / "data/raw"

train_df: pd.DataFrame = pd.read_csv(DATA_DIR / 'train.csv')
val_df: pd.DataFrame = pd.read_csv(DATA_DIR / 'val.csv')

classes_df: pd.DataFrame = pd.read_csv(DATA_DIR / 'classes.csv')
```

```python
def show_image(path):
    image = Image.open(path)
    plt.imshow(image)
    plt.axis("off")
    plt.show()
```

```python
n = 451

print(f"Image ID: {train_df.iloc[n]['id']}")
print(f"Label: {train_df.iloc[n]['label']}")
print(f"Class Name: {classes_df[classes_df['label'] == train_df.iloc[n]['label']]['class_name'].values[0]}")
print(f"Image Shape: {Image.open(DATA_DIR / 'train' / train_df.iloc[n]['id']).size}")
show_image(DATA_DIR / "train" / train_df.iloc[n]["id"])
```

```python
sizes = []

for filename in train_df["id"]:
    with Image.open(DATA_DIR / "train" / filename) as img:
        sizes.append(img.size)

pd.Series(sizes).value_counts().head(20)
```

```python
sizes = []

for filename in val_df["id"]:
    with Image.open(DATA_DIR / "val" / filename) as img:
        sizes.append(img.size)

pd.Series(sizes).value_counts().head(20)
```

```python
modes = []

for filename in train_df["id"]:
    with Image.open(DATA_DIR / "train" / filename) as img:
        modes.append(img.mode)

pd.Series(modes).value_counts()
```

```python
modes = []

for filename in val_df["id"]:
    with Image.open(DATA_DIR / "val" / filename) as img:
        modes.append(img.mode)

pd.Series(modes).value_counts()
```

```python
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = REPOSITORY_ROOT / "data/raw"

val_df = pd.read_csv(DATA_DIR / "val.csv")
sample = val_df.sample(n=100, random_state=None)

fig, axes = plt.subplots(10, 10, figsize=(18, 18))

for ax, (_, row) in zip(axes.ravel(), sample.iterrows()):
    image = Image.open(DATA_DIR / "val" / row["id"]).convert("RGB")

    ax.imshow(image)
    # ax.set_title(f'{row["label"]}', fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
